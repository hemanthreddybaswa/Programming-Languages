```python
#identifiers - A Python identifier is a name used to identify a variable, function, class, 
# module or other object.
'''
Here are naming conventions for Python identifiers:
1.Class names start with an uppercase letter. All other identifiers start with a 
lowercase letter.
2.Starting an identifier with a single leading underscore indicates that the identifier 
is private.
3.Starting an identifier with two leading underscores indicates a strong 
private identifier.
4.If the identifier also ends with two trailing underscores, the identifier is a 
language-defined special name.
'''
number=10
name="Thomas"
print(number)
print(name)
```

```python
#keywords - These are reserved words and you cannot use them as constants
# or variables or any other identifier names
# eg: if else for while from import ....
if 5>3:
    print("True")
```

```python
#Lines and Indentation
'''
Python does not use braces({}) to indicate blocks of code for class and function definitions or
flow control. Blocks of code are denoted by line indentation, which is rigidly enforced.
The number of spaces in the indentation is variable, but all statements within the block 
must be indented the same amount. 
'''
if 5>3:
    print ("True")
else:
    print ("False")    
```

```python
#Multi line statements uses \
'''
Multi-Line Statements
Statements in Python typically end with a new line. Python, however, allows the use of 
the line continuation character (\') to denote that the line should continue. For example
total = item_one + \'
item_two + \'
item_three
The statements contained within the [], {}, or () brackets do not need to use the line 
continuation character. For example
days = ['Monday', 'Tuesday', 'Wednesday',
'Thursday', 'Friday' 
'''
print("Multi line statements")
```

```python
# Comments in python
#A hash sign (#) that is not inside a string literal is the beginning of a comment.
#Multi-line Comment: Python does not have a specific multi-line comment syntax.
#The PEP 8 Style Guide recommends using consecutive single-line comments

print("Python") #inline comment for printing python

#Alternative:  DocstringsYou can also use multi-line strings (""" or ''') as comments
#  if they are not assigned to a variable.
# Python ignores unassigned string literals, 
# though they are technically processed as data and 
# are officially meant for documentation strings

''' 
This section is about comments. 
'''
print(__doc__)

```
